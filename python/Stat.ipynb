```python
M_mean_ocean = [[],[]]
M_mean_desert = [[],[]]
M_mean_montagne = [[],[]]
M_mean_ville = [[],[]]
M_mean_foret = [[],[]]

#moy des scores TMO par scène
for k in range(len(M_ocean[0])):
    M_mean_ocean[0].append(np.mean(M_ocean[0][k])) #moy des résultats de l'eq 8
    M_mean_ocean[1].append(np.mean(M_ocean[1][k])) #moy des résultats du direct
    
    M_mean_desert[0].append(np.mean(M_desert[0][k])) 
    M_mean_desert[1].append(np.mean(M_desert[1][k]))
    
    M_mean_ville[0].append(np.mean(M_ville[0][k])) 
    M_mean_ville[1].append(np.mean(M_ville[1][k]))
    
    M_mean_montagne[0].append(np.mean(M_montagne[0][k])) 
    M_mean_montagne[1].append(np.mean(M_montagne[1][k]))
    
    M_mean_foret[0].append(np.mean(M_foret[0][k])) 
    M_mean_foret[1].append(np.mean(M_foret[1][k]))
```

```python
#moy des scores TMO en combinant sans distinction des éq par scène
def mean_eq(M_in):
    M = []
    for k in range(len(M_in[0])):
        M.append((M_in[0][k]+M_in[1][k])/2) #moy sans distinction des eq
    return M

M_mean_ocean_eq = mean_eq(M_mean_ocean)
M_mean_desert_eq = mean_eq(M_mean_desert)
M_mean_ville_eq = mean_eq(M_mean_ville)
M_mean_foret_eq = mean_eq(M_mean_foret)
M_mean_montagne_eq = mean_eq(M_mean_montagne)
```

```python
#moy des scores TMO en combinant tous les scènes
def mean_TMO(M1,M2,M3,M4,M5, deg):
    """
    Mn : matrice moyenne des résultats des 5 scènes
    deg: son numéro de colonne
    """
    
    #Si on distingue les eq 8 et les directs
    if len(M1) == 2: 
        M_eq8 = np.mean([M1[0][deg],M2[0][deg],M3[0][deg],M4[0][deg],M5[0][deg]])
        M_direct = np.mean([M1[1][deg],M2[1][deg],M3[1][deg],M4[1][deg],M5[1][deg]])
        return M_eq8,M_direct
    
    #Si on distingue pas les eq 8 et direct
    else:
        M = np.mean([M1[deg],M2[deg],M3[deg],M4[deg],M5[deg]])
        return M
    
Fattal_sans_eq = mean_TMO(M_mean_ocean_eq,M_mean_desert_eq,M_mean_ville_eq,M_mean_foret_eq,M_mean_montagne_eq, 0)
Fattal = mean_TMO(M_mean_ocean,M_mean_desert,M_mean_ville,M_mean_foret,M_mean_montagne, 0)

Kimkautz_sans_eq = mean_TMO(M_mean_ocean_eq,M_mean_desert_eq,M_mean_ville_eq,M_mean_foret_eq,M_mean_montagne_eq, 1)
Kimkautz = mean_TMO(M_mean_ocean,M_mean_desert,M_mean_ville,M_mean_foret,M_mean_montagne, 1)

Mantiuk08_sans_eq = mean_TMO(M_mean_ocean_eq,M_mean_desert_eq,M_mean_ville_eq,M_mean_foret_eq,M_mean_montagne_eq, 2)
Mantiuk08 = mean_TMO(M_mean_ocean,M_mean_desert,M_mean_ville,M_mean_foret,M_mean_montagne, 2)

Reinhard02_sans_eq = mean_TMO(M_mean_ocean_eq,M_mean_desert_eq,M_mean_ville_eq,M_mean_foret_eq,M_mean_montagne_eq, 3)
Reinhard02 = mean_TMO(M_mean_ocean,M_mean_desert,M_mean_ville,M_mean_foret,M_mean_montagne, 3)

gamma_sans_eq = mean_TMO(M_mean_ocean_eq,M_mean_desert_eq,M_mean_ville_eq,M_mean_foret_eq,M_mean_montagne_eq, 4)
gamma = mean_TMO(M_mean_ocean,M_mean_desert,M_mean_ville,M_mean_foret,M_mean_montagne, 4)
```

```python
#Calcul du maximun global
def maxi(M1, M2, M3, M4, M5):
    nom = ["Fattal", "Kimkautz", "Mantiuk08", "Reihard02", "gamma"]
    try:
        if len(M1)==2:
            max_eq = max([M1[0],M2[0],M3[0],M4[0],M5[0]]) #meilleur algo selon eq8
            max_direct = max([M1[1],M2[1],M3[1],M4[1],M5[1]]) #meilleur algo selon direct

            deg_eq = [M1[0],M2[0],M3[0],M4[0],M5[0]].index(max_eq)
            deg_direct = [M1[1],M2[1],M3[1],M4[1],M5[1]].index(max_direct)
            TMO_eq = nom[deg_eq]
            TMO_direct = nom[deg_direct]

            return max_eq, max_direct, TMO_eq, TMO_direct
    except:
        liste = [M1, M2, M3, M4, M5]
        maxi = max(liste)
        
        deg = liste.index(maxi)
        TMO = nom[deg]
        return maxi, TMO

max_eq, max_direct, TMO_eq, TMO_direct = maxi(Fattal, Kimkautz, Mantiuk08, Reinhard02, gamma)
maxi, TMO = maxi(Fattal_sans_eq, Kimkautz_sans_eq, Mantiuk08_sans_eq, Reinhard02_sans_eq, gamma_sans_eq)

```

```python
#Calcul du maximun local (par scène)
def max_scene(M):
    nom = ["Fattal", "Kimkautz", "Mantiuk08", "Reihard02", "gamma"]
    if len(M)==2:
        Max_eq = max(M[0])
        Max = max(M[1])
        
        deg_eq = M[0].index(Max_eq)
        deg_direct = M[1].index(Max)
        
        TMO_eq = nom[deg_eq]
        TMO_direct = nom[deg_direct]
        
        return Max_eq, Max, TMO_eq, TMO_direct
    else:
        maxi = max(M)
        deg = M.index(maxi)
        TMO = nom[deg]
        return maxi, TMO
#Ocean      
Max_ocean_eq, Max_ocean_direct, TMO_eq_ocean, TMO_eq_ocean = max_scene(M_mean_ocean)
Max_ocean, TMO_ocean = max_scene(M_mean_ocean_eq)

#Desert
Max_desert_eq, Max_desert_direct, TMO_eq_desert, TMO_ddesert_irect = max_scene(M_mean_desert)
Max_desert, TMO_desert = max_scene(M_mean_desert_eq)

#Ville
Max_ville_eq, Max_ville_direct, TMO_eq_ville, TMO_ville_direct = max_scene(M_mean_ville)
Max_ville, TMO_ville = max_scene(M_mean_ville_eq)

#Foret
Max_foret_eq, Max_foret_direct, TMO_eq_foret, TMO_foret_direct = max_scene(M_mean_foret)
Max_foret, TMO_foret = max_scene(M_mean_foret_eq)

#Montagne
Max_montagne_eq, Max_montagne_direct, TMO_eq_montagne, TMO_montagne_direct = max_scene(M_mean_montagne)
Max_montagne, TMO_montagne = max_scene(M_mean_montagne_eq)
```
